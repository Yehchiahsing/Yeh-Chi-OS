````
import numpy as np
import math
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# ============================================
# AIHC Prototype Lab
# Quantum + Cryptanalysis + Machine Learning
# ============================================

# 1. Quantum Random Bit Emulator (Simulating Quantum Superposition via H-Gate)
def generate_quantum_data(n_samples=100):
    # Simulate the measurements of 1 Qubit after applying a Hadamard gate (50% 0, 50% 1)
    rng = np.random.default_rng()
    data = rng.integers(0, 2, size=n_samples)
    return data

# 2. Cryptanalysis Simulator (Comparing Classical vs. Quantum Attacks)
def password_space(alphabet, length):
    return alphabet ** length

def classical_crack_time(space, rate):
    return space / rate

def quantum_crack_time(space, rate):
    # Grover's search algorithm scales as O(sqrt(N))
    return math.sqrt(space) / rate

def simulate_attack(alphabet=36, length=12, rate=10**9):
    space = password_space(alphabet, length)
    classical_time_sec = classical_crack_time(space, rate)
    quantum_time_sec = quantum_crack_time(space, rate)

    print("\n=== AIHC Cryptanalysis Simulation ===")
    print(f"字符基值 (Alphabet Size): {alphabet}")
    print(f"密碼長度 (Password Length): {length}")
    print(f"密碼空間 (Keyspace): {space:.4e}")
    print(f"破解速率 (Tries per Sec): {rate:.1e}")
    print(f"經典暴力破解時間: {classical_time_sec / (365*24*3600):.2f} 年")
    print(f"量子 Grover 搜尋破解時間: {quantum_time_sec:.4f} 秒")

# 3. Machine Learning Security Analyzer
def run_ml_analysis():
    # Generate simulated quantum random noise as security features
    X = generate_quantum_data(200).reshape(-1, 1)
    # Simulating random labels representing binary patterns
    y = np.random.randint(0, 2, size=(200,))

    X_train, X_test = X[:150], X[150:]
    y_train, y_test = y[:150], y[150:]

    model = LogisticRegression()
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)

    print("\n=== AIHC Machine Learning Security Analyzer ===")
    print(f"模型驗證準確率 (Model Accuracy): {acc:.4f} (由於資料隨機，接近平衡隨機分佈 0.50)")

# 4. Main Console for Execution
def main():
    print("===== AIHC Prototype System Initialized =====")
    print("口令：創決置文，倉頡複製 (Decision made, Document copied by Cangjie)")

    # Run simulation
    simulate_attack(alphabet=36, length=12, rate=10**9)
    run_ml_analysis()

if __name__ == '__main__':
    main()
````

```markdown
## 五、 AIHC 實戰延伸：量子隨機加密與 LSB 圖片隱寫術系統

本單元實作如何將 AIHC 的量子隨機性應用於數據安全：
1. **Quantum XOR Encryption**: 使用量子模擬隨機位元生成一組密鑰，對明文進行 XOR 加密。
2. **LSB Steganography**: 將加密後的密文隱藏至一張圖片中（隱寫），並能完美提取還原。
```

````
import numpy as np
from PIL import Image
import io

# ==========================================
# 1. 量子隨機流加密 (Quantum XOR Encryption)
# ==========================================

def get_quantum_key_stream(length):
    """模擬量子 Hadamard 閘生成隨機密鑰流"""
    rng = np.random.default_rng()
    # 產生 0 與 1 的隨機位元陣列
    return rng.integers(0, 2, size=length, dtype=np.uint8)

def encrypt_decrypt_xor(data_bytes, key_stream):
    """使用 XOR 進行加密或解密（一次性密碼本原理）"""
    data_bin = np.unpackbits(np.frombuffer(data_bytes, dtype=np.uint8))
    # 將密鑰流與數據流進行位元 XOR 運算
    cipher_bin = data_bin ^ key_stream[:len(data_bin)]
    return np.packbits(cipher_bin).tobytes()

# ==========================================
# 2. 最低有效位隱寫術 (LSB Steganography)
# ==========================================

def embed_message(image_path, secret_bytes, output_path):
    """將二進位數據嵌入到圖像的 LSB 中"""
    img = Image.open(image_path).convert('RGB')
    pixels = np.array(img)

    # 將長度資訊與數據結合 (前 32 位元記錄數據長度)
    data_len = len(secret_bytes)
    len_bin = np.unpackbits(np.array([data_len], dtype=np.uint32).view(np.uint8))
    data_bin = np.unpackbits(np.frombuffer(secret_bytes, dtype=np.uint8))
    payload = np.concatenate([len_bin, data_bin])

    flat_pixels = pixels.flatten()
    if len(payload) > len(flat_pixels):
        raise ValueError("密文過大，無法裝入此圖片中！")

    # 使用 254 (0xFE) 替代 ~1，確保 uint8 運算不發生負數越界
    flat_pixels[:len(payload)] = (flat_pixels[:len(payload)] & 254) | payload

    # 重組並保存圖片
    encoded_pixels = flat_pixels.reshape(pixels.shape)
    encoded_img = Image.fromarray(encoded_pixels.astype(np.uint8))
    encoded_img.save(output_path)
    print(f"[成功] 密文已隱寫至圖片並匯出為: {output_path}")

def extract_message(image_path):
    """從隱寫圖像中提取被 LSB 隱藏的數據"""
    img = Image.open(image_path).convert('RGB')
    pixels = np.array(img)
    flat_pixels = pixels.flatten()

    # 1. 先提取 32 位元長度資訊
    len_bin = flat_pixels[:32] & 1
    data_len = np.packbits(len_bin).view(np.uint32)[0]

    # 2. 根據長度提取密文數據
    payload_bin = flat_pixels[32:32 + (data_len * 8)] & 1
    extracted_bytes = np.packbits(payload_bin).tobytes()
    return extracted_bytes
````

````
# ==========================================
# 3. 系統實戰演示 (Demo Console)
# ==========================================

# 建立一張隨機測試圖片 (作為隱寫載體)
img_data = np.random.randint(0, 255, (100, 100, 3), dtype=np.uint8)
cover_img = Image.fromarray(img_data)
cover_img.save('cover.png')

# 步驟 A: 準備明文
plaintext = "AIHC_Decision_Made_Cangjie_Copied_2025".encode('utf-8')
print(f"原始明文: {plaintext.decode('utf-8')}")

# 步驟 B: 生成量子隨機金鑰流並加密
bit_len = len(plaintext) * 8
quantum_key = get_quantum_key_stream(bit_len)
ciphertext = encrypt_decrypt_xor(plaintext, quantum_key)
print(f"加密後密文 (16進位): {ciphertext.hex()}")

# 步驟 C: 將密文隱藏進 cover.png 中
embed_message('cover.png', ciphertext, 'stego_image.png')

# 步驟 D: 從 stego_image.png 中還原提取
extracted_ciphertext = extract_message('stego_image.png')
print(f"從圖片中提取出的密文: {extracted_ciphertext.hex()}")

# 步驟 E: 使用相同的量子金鑰解密
decrypted_text = encrypt_decrypt_xor(extracted_ciphertext, quantum_key)
print(f"解密後明文: {decrypted_text.decode('utf-8')}")
````
